   # top 10 highest rated phones on flipkart in india

```python
highest_rated= df.sort_values(by=['Star Rating'] ,ascending=False)
highest_rated=highest_rated.head(10)
print(highest_rated['Product Name'])
```

```python
df1=df.copy()
```

# product with the highest rating

```python

labels=highest_rated['Product Name']
counts=highest_rated['Number Of Ratings']
plt.bar(labels,counts)
plt.title('highest rated iphones')
plt.xlabel("product name")
plt.ylabel("number of ratings")
plt.xticks(rotation=90)
plt.show()
```

# iphone with the highest no of reviews on flipkart

```python
labels=highest_rated['Product Name']
counts=highest_rated['Number Of Reviews']
plt.bar(labels,counts)
plt.title('most talked iphones')
plt.xlabel("product name")
plt.ylabel("number of reviews")
plt.xticks(rotation=90)
plt.show()
```

# relationship between sale price and no of ratings on flipkart

```python
plt.scatter(df['Sale Price'],df['Number Of Ratings'])
plt.xlabel("sale price")
plt.ylabel("number of ratings")
plt.title("sale price vs no of ratings comparison")
plt.show()
```

# relationship between discount percentage and no of ratings on flipkart

```python
plt.scatter(df['Discount Percentage'],df['Number Of Ratings'])
plt.xlabel("discount percentage")
plt.ylabel("number of ratings")
plt.title("discount percentage  vs no of ratings comparison")
plt.show()
```
